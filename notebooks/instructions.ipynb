### 4. Enjoy the Simulation:
5 Robots drive in a V-Shape Flocking in the direction of a global red light. The single robot on the tip of the V has the Leader role of the formation. The two robots behind the leader have the Deputy role and all following robots receive the Follower role. In case the leader dies, one of the deputies overtakes the role of the leader for the formation and an other follower robot becomes the new second deputy. 
The robots are started one after the other and a Stopper robot hold the first leader in place by a special condition to avoid losing the other right at the start.
